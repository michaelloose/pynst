# Exploring MultiIndex data

The selector and plotting helper share the same pandas representation. Selection remains usable in scripts and tests; widgets are an optional notebook adapter.

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from pynst.visualization import MultiIndexSelector
from pynst.visualization.matplotlib import plot_mi
````

````python
columns = pd.MultiIndex.from_product(
    [["gain", "efficiency"], [1.0, 2.0]],
    names=["quantity", "frequency_GHz"],
)
frame = pd.DataFrame(
    np.array([
        [10.0, 11.0, 0.40, 0.42],
        [12.0, 13.0, 0.50, 0.53],
        [14.0, 15.0, 0.55, 0.58],
    ]),
    index=pd.Index([0.0, 0.5, 1.0], name="input_power_W"),
    columns=columns,
)
frame
````

A directly supplied DataFrame uses its outer column level as the variable mapping, matching the historical interactive plotter. The remaining levels become selectors.

````python
selector = MultiIndexSelector(frame)
selected = selector.select(frequency_GHz=2.0)
selected["gain"]
````

````python
fig, ax = plt.subplots(figsize=(6, 3.5))
plot_mi(
    selector.dataframes["gain"],
    ax=ax,
    label_levels=["frequency_GHz"],
)
ax.set(xlabel="Input power (W)", ylabel="Gain (dB)")
ax.grid(True)
ax.legend()
fig.tight_layout()
fig
````

## Adding notebook controls

In Jupyter, install `pynst[visualization]` and explicitly display the widget:

```python
from pynst.visualization import InteractiveMultiIndexPlotter

plotter = InteractiveMultiIndexPlotter(
    frame,
    active_levels=["frequency_GHz"],
)
plotter.show()
```

Use `plotter.widget()` when another notebook layout should own the display operation.
